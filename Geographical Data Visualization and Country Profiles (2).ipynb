```python
import folium
import pandas as pd

# Country Profiles
countries = pd.DataFrame({
    'Country': ['South Africa', 'DRC (Congo)', 'Namibia', 'Mozambique'],
    'Latitude': [-30.5595, -2.8, -22.6, -18.6],
    'Longitude': [22.9375, 23.7, 17.1, 35.5],
    'GDP_BillionUSD': [350, 55, 15, 20],
    'MiningRevenue_BillionUSD': [25, 12, 3, 4],
})

# Major Mineral Deposits
deposits = pd.DataFrame({
    'Country': ['South Africa', 'South Africa', 'DRC (Congo)', 'Namibia', 'Mozambique'],
    'DepositName': ['Waterberg Lithium Mine', 'Waterberg Lithium Mine', 'Kolwezi Mine', 'Otjozondu Manganese', 'Balama Graphite Mine'],
    'Mineral': ['Lithium', 'Lithium', 'Cobalt', 'Manganese', 'Graphite'],
    'Latitude': [-24.5, -24.5, -10.7, -22.0, -13.3],
    'Longitude': [28.5, 28.5, 25.5, 17.2, 38.2],
    'Production_tonnes': [40000, 40000, 95000, 30000, 12000]
})

# Mineral Color Mapping
mineral_colors = {
    'Lithium': 'blue',
    'Cobalt': 'darkblue',
    'Manganese': 'brown',
    'Graphite': 'gray'
}

# Initialize Map
map_center = [-10, 25]
m = folium.Map(location=map_center, zoom_start=3, tiles='cartodb positron')

# Add Country Markers
for _, row in countries.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=f"<h4>{row['Country']}</h4><b>GDP:</b> {row['GDP_BillionUSD']} Billion USD<br><b>Mining Revenue:</b> {row['MiningRevenue_BillionUSD']} Billion USD",
        icon=folium.Icon(color='darkblue', icon='flag')
    ).add_to(m)

# Add Mineral Deposit Markers
for _, row in deposits.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=6,
        color=mineral_colors.get(row['Mineral'], 'orange'),
        fill=True,
        fill_color=mineral_colors.get(row['Mineral'], 'orange'),
        fill_opacity=0.7,
        popup=f"<b>Deposit:</b> {row['DepositName']}<br><b>Mineral:</b> {row['Mineral']}<br><b>Production:</b> {row['Production_tonnes']} tonnes"
    ).add_to(m)

# Add Mineral Legend
legend_html = '''
    <div style="position: absolute; bottom: 10px; left: 10px; width: 180px; height: 130px;
                background-color: white; opacity: 0.8; padding: 10px; font-size: 12px; border-radius: 5px;
                z-index: 1000; border: 2px solid grey;">
        <h4>Mineral Legend</h4>
        <ul style="list-style-type: none; padding: 0;">
            <li><i style="background-color: blue; padding: 5px;"></i> Lithium</li>
            <li><i style="background-color: darkblue; padding: 5px;"></i> Cobalt</li>
            <li><i style="background-color: brown; padding: 5px;"></i> Manganese</li>
            <li><i style="background-color: gray; padding: 5px;"></i> Graphite</li>
        </ul>
    </div>
'''

folium.Marker(
    location=map_center,
    icon=folium.DivIcon(html=legend_html)
).add_to(m)

# Display map
m
```
